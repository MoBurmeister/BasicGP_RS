```python
from scipy.spatial import Delaunay
import plotly.graph_objects as go

# Create a 3D scatter plot for all points, colored by index
scatter_points1 = go.Scatter3d(
    x=df_output_data.iloc[:, 0],  # Objective 1
    y=df_output_data.iloc[:, 1],  # Objective 2
    z=df_output_data.iloc[:, 2],  # Objective 3
    mode='markers',
    marker=dict(
        size=7,
        color=df_output_data.index,  # Use the index for color mapping
        colorscale='Viridis',  # Use Viridis colormap
        showscale=True,  # Show color bar
        colorbar=dict(
            title=dict(text='Iteration', font=dict(size=32, family='Arial, bold')),
            tickfont=dict(size=20, family='Arial'),  # Set color scale number font size
            len=0.5,  # Set color bar length
            y = 0.5,  # Set color bar position
            x=1.15  # Set color bar position
        )
    ),
    name='All 3D Observations'
)


pareto_surface1 = go.Mesh3d(
    x=df_pareto_front.iloc[:, 0],
    y=df_pareto_front.iloc[:, 1],
    z=df_pareto_front.iloc[:, 2],
    color='red',
    opacity=0.2,
    name='Pareto Front Surface'
)

# Triangulate the Pareto points for mesh construction
points = df_pareto_front.iloc[:, :3].values  # Extract Pareto points as NumPy array
tri = Delaunay(points[:, :2])  # Triangulate on the first two dimensions (X, Y)

# Construct the Pareto surface with triangulation
pareto_surface_mesh = go.Mesh3d(
    x=points[:, 0],
    y=points[:, 1],
    z=points[:, 2],
    i=tri.simplices[:, 0],  # Triangulation indices for first vertex
    j=tri.simplices[:, 1],  # Triangulation indices for second vertex
    k=tri.simplices[:, 2],  # Triangulation indices for third vertex
    color='red',
    opacity=0.3,
    name='Pareto Front Mesh'
)


# Add visible edges as a line trace
edges_x = []
edges_y = []
edges_z = []

for simplex in tri.simplices:
    # Extract the vertices of the triangle
    for i, j in [(0, 1), (1, 2), (2, 0)]:  # Edges: (0->1, 1->2, 2->0)
        edges_x += [points[simplex[i], 0], points[simplex[j], 0], None]  # X-coords
        edges_y += [points[simplex[i], 1], points[simplex[j], 1], None]  # Y-coords
        edges_z += [points[simplex[i], 2], points[simplex[j], 2], None]  # Z-coords

pareto_edges = go.Scatter3d(
    x=edges_x,
    y=edges_y,
    z=edges_z,
    mode='lines',
    line=dict(color='black', width=2),
    name='Pareto Surface Edges'
)


# Manually set axis ranges
x_range1 = [1625, 1720]  # Replace with your desired range for X-axis
y_range1 = [1.5, 14.5]   # Replace with your desired range for Y-axis
z_range1 = [-0.06, 0.3] # Replace with your desired range for Z-axis


# Scatter projections on the X-Z plane (Y constant) for all points
scatter_xz = go.Scatter3d(
    x=df_output_data.iloc[:, 0],  # Objective 1
    y=[y_range1[0]] * len(df_output_data),  # Y is fixed
    z=df_output_data.iloc[:, 2],  # Objective 3
    mode='markers',
    marker=dict(
        size=4,
        color='darkgray',  # Set to dark gray
        opacity=0.99  # Make projections semi-transparent
    ),
    name='Projection X-Z (All Points)',
    showlegend=False
)

# Scatter projections on the Y-Z plane (X constant) for all points
scatter_yz = go.Scatter3d(
    x=[x_range1[0]] * len(df_output_data),  # X is fixed
    y=df_output_data.iloc[:, 1],  # Objective 2
    z=df_output_data.iloc[:, 2],  # Objective 3
    mode='markers',
    marker=dict(
        size=4,
        color='darkgray',  # Set to dark gray
        opacity=0.99  # Make projections semi-transparent
    ),
    name='Projection Y-Z (All Points)',
    showlegend=False
)

# Scatter projections on the X-Y plane (Z constant) for all points
scatter_xy = go.Scatter3d(
    x=df_output_data.iloc[:, 0],  # Objective 1
    y=df_output_data.iloc[:, 1],  # Objective 2
    z=[z_range1[0]] * len(df_output_data),  # Z is fixed
    mode='markers',
    marker=dict(
        size=4,
        color='darkgray',  # Set to dark gray
        opacity=0.99  # Make projections semi-transparent
    ),
    name='Projections of All Points'
)

# Pareto front projection on the X-Z plane (Y constant)
pareto_xz = go.Scatter3d(
    x=df_pareto_front.iloc[:, 0],  # Objective 1
    y=[y_range1[0]] * len(df_pareto_front),  # Y is fixed
    z=df_pareto_front.iloc[:, 2],  # Objective 3
    mode='markers',
    marker=dict(
        size=4,
        color='black',  # Set to dark red
        opacity=0.99
    ),
    name='Pareto X-Z',
    showlegend=False
)

# Pareto front projection on the Y-Z plane (X constant)
pareto_yz = go.Scatter3d(
    x=[x_range1[0]] * len(df_pareto_front),  # X is fixed
    y=df_pareto_front.iloc[:, 1],  # Objective 2
    z=df_pareto_front.iloc[:, 2],  # Objective 3
    mode='markers',
    marker=dict(
        size=4,
        color='black',  # Set to dark red
        opacity=0.99
    ),
    name='Pareto Y-Z',
    showlegend=False
)

# Pareto front projection on the X-Y plane (Z constant)
pareto_xy = go.Scatter3d(
    x=df_pareto_front.iloc[:, 0],  # Objective 1
    y=df_pareto_front.iloc[:, 1],  # Objective 2
    z=[z_range1[0]] * len(df_pareto_front),  # Z is fixed
    mode='markers',
    marker=dict(
        size=4,
        color='black',  # Set to dark red
        opacity=0.99
    ),
    name='Projections of Pareto Points'
)

# Combine all the plots (all points and Pareto front projections)
fig = go.Figure(data=[   scatter_points1, pareto_surface_mesh,
    scatter_xz, scatter_yz, scatter_xy,  # All points in gray
    pareto_xz, pareto_yz, pareto_xy      # Pareto points in dark red
])

# Set the layout as before
fig.update_layout(
    legend=dict(
        x=1.05,
        y=0.9,
        font=dict(
            size=20,
            family='Arial'
        ),
        bgcolor='rgba(255,255,255,0.7)',
        bordercolor='black',
        borderwidth=1
    ),
    scene=dict(
        xaxis=dict(
            title=dict(text='Mass [kg]', font=dict(size=28, family='Arial, bold')),
            range=x_range1,
            tickfont=dict(size=16, family='Arial'),
            dtick=20
        ),
        yaxis=dict(
            title=dict(text='A_in [m/s]', font=dict(size=28, family='Arial, bold')),
            range=y_range1,
            tickfont=dict(size=16, family='Arial')
        ),
        zaxis=dict(
            title=dict(text='Intrusion [m]', font=dict(size=28, family='Arial, bold')),
            range=z_range1,
            tickfont=dict(size=16, family='Arial')
        ),
        camera=dict(
            eye=dict(x=1.5, y=1.5, z=0.85),
            up=dict(x=0, y=0, z=1),
        ),
    ),
    title=dict(
        text='3D Pareto Front with Projections "No TL"',
        font=dict(size=50, family='Arial, bold'),
        x=0.5,
        y=0.94,
        xanchor='center'
    ),
    width=1200,
    height=1080,
)

# Show the interactive plot
fig.show()
```

```python
def plot_scatter_with_colorbar(df, title, x_label, y_label, width_cm=25, height_cm=20):
    """
    Creates a scatter plot from a dataframe with specified figure dimensions and labels.
    
    Parameters:
    df (pd.DataFrame): The dataframe containing the data to plot. The first two columns will be used.
    title (str): Title of the plot.
    x_label (str): Label for the x-axis.
    y_label (str): Label for the y-axis.
    width_cm (float): Width of the plot in centimeters.
    height_cm (float): Height of the plot in centimeters.
    
    Returns:
    None: Displays the plot.
    """
    # Convert centimeters to inches
    cm_to_inch = 2.54
    width_inch = width_cm / cm_to_inch
    height_inch = height_cm / cm_to_inch

    # Set the font globally for the plot
    plt.rcParams["font.family"] = "Arial"

    # Set the figure size (width, height) in inches
    plt.figure(figsize=(width_inch, height_inch))

    # Create a scatter plot with point size and border (edge) color
    plt.scatter(df.iloc[:, 0], df.iloc[:, 1], c=df.index, cmap="viridis", s=80, edgecolors='black', linewidths=0)

    # Set title and labels for the axes
    plt.title(title, fontname='Arial', fontsize=16, fontweight='bold')
    plt.xlabel(x_label, fontname='Arial', fontsize=11)
    plt.ylabel(y_label, fontname='Arial', fontsize=11)

    # Add a colorbar to represent the index-based color mapping
    cbar = plt.colorbar(label='Iteration')
    cbar.set_label('Iteration', fontsize=11, fontname='Arial')

    # Adjust the font size of the tick labels on the x and y axes
    plt.tick_params(axis='both', which='major', labelsize=10)

    # Add a legend below the graph
    plt.legend(
        ['Observations'],  # Add your legend labels here
        loc='upper center',
        bbox_to_anchor=(0.5, -0.1),  # Position the legend below the plot
        ncol=1,  # Adjust for the number of legend entries
        fontsize=10,
        facecolor='white'
    )

    # Show the plot
    plt.tight_layout()  # Adjust layout to ensure everything fits
    plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

def plot_hypervolume_with_rate_of_change(df, title, x_label, y_label_hv, y_label_rate, y_label_acq, width_cm=25, height_cm=20):
    """
    Creates a plot for hypervolume evolution, its rate of change, and acquisition value with iterations.
    
    Parameters:
    df (pd.DataFrame): The dataframe containing the hypervolume data, rate of change, and acquisition value.
                       Assumes 'hypervolume', 'hypervolume_rate_of_change', and 'acq_value' columns exist.
    title (str): Title of the plot.
    x_label (str): Label for the x-axis.
    y_label_hv (str): Label for the primary y-axis (hypervolume).
    y_label_rate (str): Label for the secondary y-axis (rate of change).
    y_label_acq (str): Label for the tertiary y-axis (acquisition value).
    width_cm (float): Width of the plot in centimeters.
    height_cm (float): Height of the plot in centimeters.
    
    Returns:
    None: Displays the plot.
    """
    # Convert centimeters to inches
    cm_to_inch = 2.54
    width_inch = width_cm / cm_to_inch
    height_inch = height_cm / cm_to_inch

    # Set the font globally for the plot
    plt.rcParams["font.family"] = "Arial"

    # Set the figure size (width, height) in inches
    fig, ax1 = plt.subplots(figsize=(width_inch, height_inch))

    # Plot the hypervolume evolution on the primary y-axis
    ax1.plot(
        df.index, df['hypervolume'], marker='o', linestyle='-', color='#22A884', label='Hypervolume',
        linewidth=6, markersize=6, zorder=2
    )
    ax1.scatter(
        df.index, df['hypervolume'], s=70, linewidths=0.5, zorder=3, color='#22A884', label='HV Points'
    )

    # Customize primary y-axis (hypervolume)
    ax1.set_xlabel(x_label, fontname='Arial', fontsize=20)
    ax1.set_ylabel(y_label_hv, fontname='Arial', fontsize=20)
    ax1.tick_params(axis='y', labelsize=20)
    ax1.tick_params(axis='x', labelsize=20)
    ax1.set_ylim((40, 245))

    # Create a secondary y-axis for the rate of change
    ax2 = ax1.twinx()
    ax2.plot(
        df.index, df['hypervolume_rateofchange'], linestyle='--', color='#4682B4', label='HV Rate of Change', linewidth=6, zorder=1
    )

    # Customize secondary y-axis (rate of change)
    ax2.set_ylabel(y_label_rate, fontname='Arial', fontsize=20)
    ax2.tick_params(axis='y', labelsize=20)
    ax2.yaxis.set_major_locator(MaxNLocator(integer=True))
    ax2.set_ylim((0, 220))

    # Create a tertiary y-axis for acquisition value
    ax3 = ax1.twinx()
    ax3.spines["right"].set_position(("outward", 60))  # Offset for the tertiary axis
    ax3.plot(
        df.index, df['acq_value'], linestyle='-.', color='gray', label='Acquisition Value', linewidth=6, zorder=1
    )

    # Customize tertiary y-axis (acquisition value)
    ax3.set_ylabel(y_label_acq, fontname='Arial', fontsize=20)
    ax3.tick_params(axis='y', labelsize=20)
    ax3.set_ylim((0, 10))

    # Combine legends from all axes
    lines_1, labels_1 = ax1.get_legend_handles_labels()
    lines_2, labels_2 = ax2.get_legend_handles_labels()
    lines_3, labels_3 = ax3.get_legend_handles_labels()
    combined_legend = ax1.legend(
        lines_1 + lines_2 + lines_3, labels_1 + labels_2 + labels_3, loc='upper center',
        bbox_to_anchor=(0.5, -0.1), ncol=2, facecolor='white', fontsize=20
    )
    

    # Set the title
    plt.title(title, fontname='Arial', fontsize=26, fontweight='bold', pad=20)
    # Adjust layout to ensure everything fits
    plt.tight_layout()

    # Show the plot
    plt.show()
```

```python
import pandas as pd
import re

# Define a function to extract both values from the tensor string
def extract_values(tensor_string):
    # Use regular expressions to find the numbers inside the brackets
    match = re.search(r'\[([0-9.,\s]+)\]', tensor_string)
    if match:
        # Split the extracted string by commas and return the values as floats
        values = match.group(1).split(',')
        return float(values[0].strip()), float(values[1].strip())
    return None, None

# Apply the extraction function to the column and expand the result into two new columns
df_extracted = df_output_data['diversity_metric_list'].apply(extract_values).apply(pd.Series)

# Rename the columns for clarity
df_extracted.columns = ['value_1', 'value_2']

# Display the new DataFrame
print(df_extracted)
```

```python
import matplotlib.pyplot as plt

def plot_two_columns(df, title, x_label, y_label_1, y_label_2, width_cm=25, height_cm=20):
    """
    Creates a plot for two columns from a dataframe.

    Parameters:
    df (pd.DataFrame): The dataframe containing the data for the two columns to plot.
                       Assumes 'value_1' and 'value_2' columns exist.
    title (str): Title of the plot.
    x_label (str): Label for the x-axis.
    y_label_1 (str): Label for the primary y-axis (value_1).
    y_label_2 (str): Label for the secondary y-axis (value_2).
    width_cm (float): Width of the plot in centimeters.
    height_cm (float): Height of the plot in centimeters.

    Returns:
    None: Displays the plot.
    """
    # Convert centimeters to inches
    cm_to_inch = 2.54
    width_inch = width_cm / cm_to_inch
    height_inch = height_cm / cm_to_inch

    # Set the font globally for the plot
    plt.rcParams["font.family"] = "Arial"

    # Set the figure size (width, height) in inches
    fig, ax1 = plt.subplots(figsize=(width_inch, height_inch))

    # Plot value_1 on the primary y-axis
    ax1.plot(
        df.index, df['value_1'], marker='o', linestyle='-', color='#22A884', label='Value 1',
        linewidth=3, markersize=6, zorder=2
    )
    ax1.scatter(
        df.index, df['value_1'], s=70, edgecolors='black', linewidths=0.5, zorder=3, color='#22A884', label='Value 1 Points'
    )

    # Customize primary y-axis (value_1)
    ax1.set_xlabel(x_label, fontname='Arial', fontsize=11)
    ax1.set_ylabel(y_label_1, fontname='Arial', fontsize=11)
    ax1.tick_params(axis='y', labelsize=10)
    ax1.tick_params(axis='x', labelsize=10)

    # Create a secondary y-axis for value_2
    ax2 = ax1.twinx()
    ax2.plot(
        df.index, df['value_2'], color='#4682B4', label='Value 2', linewidth=3, zorder=1
    )
    ax2.scatter(
        df.index, df['value_2'], s=70, edgecolors='black', linewidths=0.5, zorder=3, color='#4682B4', label='Value 2 Points'
    )

    # Customize secondary y-axis (value_2)
    ax2.set_ylabel(y_label_2, fontname='Arial', fontsize=11)
    ax2.tick_params(axis='y', labelsize=10)

    # Combine legends from both axes
    lines_1, labels_1 = ax1.get_legend_handles_labels()
    lines_2, labels_2 = ax2.get_legend_handles_labels()
    combined_legend = ax1.legend(
        lines_1 + lines_2, labels_1 + labels_2, loc='upper center',
        bbox_to_anchor=(0.5, -0.08), ncol=len(lines_1 + lines_2), facecolor='white', fontsize=10
    )

    # Set the title
    plt.title(title, fontname='Arial', fontsize=14)
    # Adjust layout to ensure everything fits
    plt.tight_layout()

    # Show the plot
    plt.show()
```

```python
import numpy as np

def plot_combined_optimization_time_with_regression(df, title, x_label, y_label, width_cm=25, height_cm=25):
    """
    Plots the core_optimization_time column for all sources in a combined plot and adds linear regression lines.

    Parameters:
    df (pd.DataFrame): Combined dataframe containing core_optimization_time from all sources.
    title (str): Title of the plot.
    x_label (str): Label for the x-axis.
    y_label (str): Label for the y-axis.
    width_cm (float): Width of the plot in centimeters.
    height_cm (float): Height of the plot in centimeters.

    Returns:
    None: Displays the plot.
    """
    # Convert centimeters to inches
    cm_to_inch = 2.54
    width_inch = width_cm / cm_to_inch
    height_inch = height_cm / cm_to_inch

    # Set the font globally for the plot
    plt.rcParams["font.family"] = "Arial"

    # Set the figure size (width, height) in inches
    fig, ax = plt.subplots(figsize=(width_inch, height_inch))

    colors = {
        "No TL": "gray",
        "Finetune": "#22A884",
        "Multitask": "#4682B4",
        "RGPE": "goldenrod",
        "Weighted": "#440154"
    }

    for column, color in colors.items():
        # Plot the original data
        ax.plot(
            df.index, df[column], marker='o', linestyle='-',
            color=color, label=column, linewidth=6, markersize=6, zorder=2
        )
        
        # Perform linear regression
        x = np.arange(len(df.index))  # x-values (0, 1, 2, ...)
        y = df[column].values         # y-values
        coeffs = np.polyfit(x, y, 1)  # Fit a linear regression (degree 1)
        y_fit = np.polyval(coeffs, x) # Compute the regression line

        # Plot the regression line
        ax.plot(
            df.index, y_fit, linestyle='--', color=color, linewidth=3,
            label=f"{column} (Trend)", zorder=1
        )

    # Set the axis labels and title
    ax.set_xlabel(x_label, fontname='Arial', fontsize=20)
    ax.set_ylabel(y_label, fontname='Arial', fontsize=20)
    ax.tick_params(axis='y', labelsize=20)
    ax.tick_params(axis='x', labelsize=20)

    # Set the title
    plt.title(title, fontname='Arial', fontsize=26, fontweight='bold', pad=20)

    # Add legend
    ax.legend(
        loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=2,
        facecolor='white', fontsize=20
    )

    # Adjust layout to ensure everything fits
    plt.tight_layout()

    # Show the plot
    plt.show()
```

```python
def plot_combined_number_pareto_points(df, title, x_label, y_label, width_cm=25, height_cm=25):
    """
    Plots the core_optimization_time column for all sources in a combined plot.

    Parameters:
    df (pd.DataFrame): Combined dataframe containing core_optimization_time from all sources.
    title (str): Title of the plot.
    x_label (str): Label for the x-axis.
    y_label (str): Label for the y-axis.
    width_cm (float): Width of the plot in centimeters.
    height_cm (float): Height of the plot in centimeters.

    Returns:
    None: Displays the plot.
    """
    # Convert centimeters to inches
    cm_to_inch = 2.54
    width_inch = width_cm / cm_to_inch
    height_inch = height_cm / cm_to_inch

    # Set the font globally for the plot
    plt.rcParams["font.family"] = "Arial"

    # Set the figure size (width, height) in inches
    fig, ax = plt.subplots(figsize=(width_inch, height_inch))

    # Plot each column individually with specific colors
    ax.plot(
        df.index, df["No TL"], marker='o', linestyle='-',
        color='gray', label="No Transfer Learning", linewidth=6, markersize=6, zorder=2
    )
    ax.plot(
        df.index, df["Finetune"], marker='o', linestyle='-',
        color='#22A884', label="Finetune", linewidth=6, markersize=6, zorder=2
    )
    ax.plot(
        df.index, df["Multitask"], marker='o', linestyle='-',
        color='#4682B4', label="Multitask", linewidth=6, markersize=6, zorder=2
    )
    ax.plot(
        df.index, df["RGPE"], marker='o', linestyle='-',
        color='goldenrod', label="RGPE", linewidth=6, markersize=6, zorder=2
    )
    ax.plot(
        df.index, df["Weighted"], marker='o', linestyle='-',
        color='#440154', label="Weighted", linewidth=6, markersize=6, zorder=2
    )

    # Set the axis labels and title
    ax.set_xlabel(x_label, fontname='Arial', fontsize=20)
    ax.set_ylabel(y_label, fontname='Arial', fontsize=20)
    ax.tick_params(axis='y', labelsize=20)
    ax.tick_params(axis='x', labelsize=20)
    ax.set_ylim([0 , 250])
    ax.set_xlim([0, 60])

    # Set the title
    plt.title(title, fontname='Arial', fontsize=26, fontweight='bold', pad=20)

    # Add legend
    ax.legend(
        loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=2,
        facecolor='white', fontsize=20
    )

    # Adjust layout to ensure everything fits
    plt.tight_layout()

    # Show the plot
    plt.show()
```

```python
# Call the function to plot the data
plot_combined_number_pareto_points(
    combined_data,
    title="Vehicle Crash Hypervolume Progress Across All Strategies",
    x_label="Iteration",
    y_label="Hypervolume"
)
```
